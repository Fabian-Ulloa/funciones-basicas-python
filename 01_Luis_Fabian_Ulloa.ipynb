## 1. `print()`

**¿Qué hace?** Muestra información en pantalla.

**Sintaxis:** `print(valor1, valor2, sep=' ', end='\n')`

**Ejemplo práctico:**

```python
print("Hola", "mundo")
print("Python", "es", "genial", sep="-")
print("Sin salto de línea", end="... ")
print("continúa aquí")
```

**💡 Explicación:** `sep` cambia el separador entre valores y `end` lo que se escribe al final.

## 2. `input()`

**¿Qué hace?** Pide un dato al usuario por teclado. Siempre devuelve texto (str).

**Sintaxis:** `variable = input('mensaje')`

**Ejemplo práctico:**

```python
nombre = input("¿Cómo te llamas? ")
print("Hola,", nombre)
```

**💡 Explicación:** Aunque el usuario escriba un número, `input()` lo entrega como texto.

## 3. `type()`

**¿Qué hace?** Dice de qué tipo es un dato.

**Sintaxis:** `type(objeto)`

**Ejemplo práctico:**

```python
print(type(5))
print(type(3.14))
print(type("hola"))
print(type(True))
```

**💡 Explicación:** Tipos básicos: `int` (entero), `float` (decimal), `str` (texto), `bool` (verdadero/falso).

## 4. `int()`

**¿Qué hace?** Convierte un valor a número entero.

**Sintaxis:** `int(valor)`

**Ejemplo práctico:**

```python
texto = "25"
print(int(texto) + 5)
print(int(3.9))
```

**💡 Explicación:** Al convertir un decimal, `int()` corta la parte decimal (no redondea).

## 5. `float()`

**¿Qué hace?** Convierte un valor a número decimal.

**Sintaxis:** `float(valor)`

**Ejemplo práctico:**

```python
print(float("3.5") + 1)
print(float(7))
```

**💡 Explicación:** Útil cuando se leen números con decimales desde `input()`.

## 6. `str()`

**¿Qué hace?** Convierte un valor a texto.

**Sintaxis:** `str(valor)`

**Ejemplo práctico:**

```python
edad = 20
print("Tengo " + str(edad) + " años")
```

**💡 Explicación:** No se puede sumar texto con número directamente; primero se convierte con `str()`.

## 7. `bool()`

**¿Qué hace?** Convierte un valor a verdadero (True) o falso (False).

**Sintaxis:** `bool(valor)`

**Ejemplo práctico:**

```python
print(bool(0))
print(bool(1))
print(bool(""))
print(bool("hola"))
```

**💡 Explicación:** Son `False`: 0, texto vacío, lista vacía y `None`. Casi todo lo demás es `True`.

## 8. `len()`

**¿Qué hace?** Cuenta cuántos elementos o caracteres tiene algo.

**Sintaxis:** `len(objeto)`

**Ejemplo práctico:**

```python
print(len("Python"))
print(len([10, 20, 30, 40]))
```

**💡 Explicación:** Funciona con textos, listas, tuplas, diccionarios, etc.

## 9. `range()`

**¿Qué hace?** Genera una secuencia de números.

**Sintaxis:** `range(inicio, fin, paso)`

**Ejemplo práctico:**

```python
print(list(range(5)))
print(list(range(2, 10, 2)))
for i in range(3):
    print("Vuelta", i)
```

**💡 Explicación:** El número final **no** se incluye. Se usa mucho en ciclos `for`.

## 10. `isinstance()`

**¿Qué hace?** Verifica si un dato es de cierto tipo.

**Sintaxis:** `isinstance(objeto, tipo)`

**Ejemplo práctico:**

```python
print(isinstance(5, int))
print(isinstance("a", int))
print(isinstance(2.5, float))
```

**💡 Explicación:** Devuelve `True` o `False`.
