## 1. `abs()`

**¿Qué hace?** Devuelve el valor absoluto (el número sin signo negativo).

**Sintaxis:** `abs(numero)`

**Ejemplo práctico:**

```python
print(abs(-7))
print(abs(3.5))
```

**💡 Explicación:** La distancia de un número al cero siempre es positiva.

## 2. `round()`

**¿Qué hace?** Redondea un número.

**Sintaxis:** `round(numero, decimales)`

**Ejemplo práctico:**

```python
print(round(3.14159, 2))
print(round(2.7))
```

**💡 Explicación:** Si no se indican decimales, redondea al entero más cercano.

## 3. `pow()`

**¿Qué hace?** Calcula una potencia.

**Sintaxis:** `pow(base, exponente)`

**Ejemplo práctico:**

```python
print(pow(2, 3))
print(pow(5, 2))
```

**💡 Explicación:** Es equivalente a `2 ** 3`.

## 4. `divmod()`

**¿Qué hace?** Devuelve el cociente y el residuo de una división.

**Sintaxis:** `divmod(a, b)`

**Ejemplo práctico:**

```python
cociente, residuo = divmod(17, 5)
print("Cociente:", cociente)
print("Residuo:", residuo)
```

**💡 Explicación:** 17 dividido entre 5 da 3 y sobran 2.

## 5. `sum()`

**¿Qué hace?** Suma todos los elementos de una lista.

**Sintaxis:** `sum(lista)`

**Ejemplo práctico:**

```python
numeros = [1, 2, 3, 4]
print(sum(numeros))
```

**💡 Explicación:** Ideal para calcular totales.

## 6. `min()`

**¿Qué hace?** Devuelve el valor más pequeño.

**Sintaxis:** `min(lista)`

**Ejemplo práctico:**

```python
notas = [3.5, 4.2, 2.8, 5.0]
print(min(notas))
```

**💡 Explicación:** También funciona con textos (orden alfabético).

## 7. `max()`

**¿Qué hace?** Devuelve el valor más grande.

**Sintaxis:** `max(lista)`

**Ejemplo práctico:**

```python
notas = [3.5, 4.2, 2.8, 5.0]
print(max(notas))
```

**💡 Explicación:** Se puede usar junto con `min()` para hallar el rango de los datos.

## 8. `sorted()`

**¿Qué hace?** Ordena una lista y devuelve una nueva.

**Sintaxis:** `sorted(lista, reverse=False)`

**Ejemplo práctico:**

```python
numeros = [5, 2, 9, 1]
print(sorted(numeros))
print(sorted(numeros, reverse=True))
print(numeros)
```

**💡 Explicación:** La lista original no cambia. Con `reverse=True` ordena de mayor a menor.

## 9. `reversed()`

**¿Qué hace?** Recorre una secuencia al revés.

**Sintaxis:** `reversed(secuencia)`

**Ejemplo práctico:**

```python
numeros = [1, 2, 3]
print(list(reversed(numeros)))
```

**💡 Explicación:** Se convierte con `list()` para poder verlo.

## 10. `enumerate()`

**¿Qué hace?** Recorre una lista dando también la posición de cada elemento.

**Sintaxis:** `enumerate(lista, start=0)`

**Ejemplo práctico:**

```python
frutas = ["manzana", "pera", "uva"]
for posicion, fruta in enumerate(frutas, start=1):
    print(posicion, fruta)
```

**💡 Explicación:** Evita llevar un contador manual.
