## 1. `list()`

**¿Qué hace?** Crea una lista (colección ordenada y modificable).

**Sintaxis:** `list(iterable)`

**Ejemplo práctico:**

```python
letras = list("abc")
print(letras)
letras.append("d")
print(letras)
```

**💡 Explicación:** Convierte textos, rangos y otras colecciones en lista.

## 2. `tuple()`

**¿Qué hace?** Crea una tupla (colección ordenada que NO se puede modificar).

**Sintaxis:** `tuple(iterable)`

**Ejemplo práctico:**

```python
coordenadas = tuple([4, 7])
print(coordenadas)
print(coordenadas[0])
```

**💡 Explicación:** Sirve para datos fijos, como coordenadas.

## 3. `set()`

**¿Qué hace?** Crea un conjunto: colección sin elementos repetidos.

**Sintaxis:** `set(iterable)`

**Ejemplo práctico:**

```python
repetidos = [1, 2, 2, 3, 3, 3]
print(set(repetidos))
```

**💡 Explicación:** Muy útil para eliminar duplicados.

## 4. `dict()`

**¿Qué hace?** Crea un diccionario: datos organizados como clave → valor.

**Sintaxis:** `dict(clave=valor)`

**Ejemplo práctico:**

```python
persona = dict(nombre="Ana", edad=20)
print(persona)
print(persona["nombre"])
```

**💡 Explicación:** Se accede al valor usando su clave.

## 5. `zip()`

**¿Qué hace?** Une varias listas elemento por elemento.

**Sintaxis:** `zip(lista1, lista2)`

**Ejemplo práctico:**

```python
nombres = ["Ana", "Luis"]
edades = [20, 22]
for n, e in zip(nombres, edades):
    print(n, "tiene", e, "años")
```

**💡 Explicación:** Se detiene cuando se acaba la lista más corta.

## 6. `map()`

**¿Qué hace?** Aplica una función a cada elemento de una lista.

**Sintaxis:** `map(funcion, lista)`

**Ejemplo práctico:**

```python
texto_numeros = ["1", "2", "3"]
enteros = list(map(int, texto_numeros))
print(enteros)
```

**💡 Explicación:** Aquí convierte cada texto en un entero.

## 7. `filter()`

**¿Qué hace?** Deja solo los elementos que cumplen una condición.

**Sintaxis:** `filter(funcion, lista)`

**Ejemplo práctico:**

```python
numeros = [1, 2, 3, 4, 5, 6]
pares = list(filter(lambda n: n % 2 == 0, numeros))
print(pares)
```

**💡 Explicación:** `lambda n: ...` es una mini función de una línea.

## 8. `any()`

**¿Qué hace?** Devuelve True si AL MENOS UNO cumple la condición.

**Sintaxis:** `any(iterable)`

**Ejemplo práctico:**

```python
print(any([False, False, True]))
print(any([0, 0, 0]))
```

**💡 Explicación:** Piensa en 'O': alcanza con que uno sea verdadero.

## 9. `all()`

**¿Qué hace?** Devuelve True solo si TODOS cumplen la condición.

**Sintaxis:** `all(iterable)`

**Ejemplo práctico:**

```python
print(all([True, True, True]))
print(all([True, False, True]))
```

**💡 Explicación:** Piensa en 'Y': todos deben ser verdaderos.

## 10. `ord()`

**¿Qué hace?** Devuelve el número (código Unicode) de un carácter.

**Sintaxis:** `ord(caracter)`

**Ejemplo práctico:**

```python
print(ord("A"))
print(ord("a"))
print(chr(65))
```

**💡 Explicación:** `chr()` hace lo contrario: de número a carácter.
